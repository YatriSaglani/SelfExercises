```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("1. Find missing values using isnull")
    print("2. Count missing values column-wise.")
    print("3. Display rows containing missing values.")
    print("4. Fill numerical missing values with mean.")
    print("5. Fill categorical missing values with mode.")
    print("6. Verify that no missing values remain.")
    print("7. Exit")

    choice = input("Enter your choice : ")

    if choice == '1':
        print(df.isnull())

    elif choice == '2':
        print(df.isnull().sum())

    elif choice == '3':
        print(df[df.isnull().any(axis=1)])

    elif choice == '4':
        numerical_cols = df.select_dtypes(include=['number']).columns
        for col in numerical_cols:
            mean_value = df[col].mean()
            df[col].fillna(mean_value, inplace=True)
        print("Numerical missing values filled with mean.")

    elif choice == '5':
        categorical_cols = df.select_dtypes(include=['object']).columns
        for col in categorical_cols:
            mode_value = df[col].mode()[0]
            df[col].fillna(mode_value, inplace=True)
        print("Categorical missing values filled with mode.")

    elif choice == '6':
        if not df.isnull().any().any():
            print("No missing values remain.")
        else:
            print("There are still missing values.")

    elif choice == '7':
        print("Exiting...")
        break

    else:
        print("Invalid choice. Please try again.")
```
