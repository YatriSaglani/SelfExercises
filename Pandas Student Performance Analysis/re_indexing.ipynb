```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
   
    print("\n1. Make ID the Index")
    print("2. Rename Index to Student_ID")
    print("3. Apply Custom Index")
    print("4. Reset Index")
    print("5. Change Selected Column Order")
    print("6. Exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':
        df_indexed = df.set_index("id")

        print("\nID as Index:")
        print(df_indexed)

    elif choice == '2':
        df_indexed = df.set_index("id")
        df_indexed.index.name = "Student_ID"

        print("\nRenamed Index:")
        print(df_indexed)

    elif choice == '3':
        custom_index = [f"Student_{i}" for i in range(1, len(df) + 1)]

        df_custom = df.copy()
        df_custom.index = custom_index

        print("\nCustom Index:")
        print(df_custom)

    elif choice == '4':
        df_indexed = df.set_index("id")
        df_reset = df_indexed.reset_index()

        print("\nAfter reset_index():")
        print(df_reset)

    elif choice == '5':
        new_order = [
            "id",
            "name",
            "course",
            "city",
            "age",
            "gender",
            "maths",
            "python",
            "pandas",
            "numpy",
            "attendance",
            "joining_date",
            "percentage"
        ]

        df_reordered = df[new_order]

        print("\nChanged Column Order:")
        print(df_reordered)

    elif choice == '6':
        print("Exiting...")
        break

    else:
        print("Invalid choice. Please try again.")
```
