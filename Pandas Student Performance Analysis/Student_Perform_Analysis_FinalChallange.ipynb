```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1. Find the highest-percentage student.")
    print("2. Find the lowest-percentage student.")
    print("3. Find the course with the highest average percentage.")
    print("4. Find the city with the highest average attendance.")
    print("5. Find the course with the most students.")
    print("6. Find the student with highest Python marks.")
    print("7. Calculate average attendance.")
    print("8. Count Pass students.")
    print("9. Count Fail students.")
    print("10. Calculate course-wise average performance.")
    print("11. Calculate city-wise average performance.")
    print("12. Calculate gender-wise performance.")
    print("13. Find students performing above their course average.")
    print("14. Find the month with the most new students.")
    print("15. Display the final analysis in a clean DataFrame.")
    print("16. Exit.")

    choice = input("\nEnter your choice (1-16): ")

    if choice == '1':
        highest_percentage_student = df.loc[df['Percentage'].idxmax()]
        print("\nHighest Percentage Student:")
        print(highest_percentage_student)

    elif choice == '2':
        lowest_percentage_student = df.loc[df['Percentage'].idxmin()]
        print("\nLowest Percentage Student:")
        print(lowest_percentage_student)

    elif choice == '3':
        course_avg_percentage = df.groupby('Course')['Percentage'].mean()
        highest_avg_course = course_avg_percentage.idxmax()
        print(f"\nCourse with the highest average percentage: {highest_avg_course} ({course_avg_percentage[highest_avg_course]:.2f}%)")

    elif choice == '4':
        city_avg_attendance = df.groupby('City')['Attendance'].mean()
        highest_avg_city = city_avg_attendance.idxmax()
        print(f"\nCity with the highest average attendance: {highest_avg_city} ({city_avg_attendance[highest_avg_city]:.2f}%)")

    elif choice == '5':
        course_student_count = df['Course'].value_counts()
        most_students_course = course_student_count.idxmax()
        print(f"\nCourse with the most students: {most_students_course} ({course_student_count[most_students_course]} students)")

    elif choice == '6':
        highest_python_marks_student = df.loc[df['Python Marks'].idxmax()]
        print("\nStudent with the highest Python marks:")
        print(highest_python_marks_student)

    elif choice == '7':
        average_attendance = df['Attendance'].mean()
        print(f"\nAverage Attendance: {average_attendance:.2f}%")

    elif choice == '8':
        pass_students_count = df[df['Percentage'] >= 40].shape[0]
        print(f"\nNumber of Pass Students: {pass_students_count}")

    elif choice == '9':
        fail_students_count = df[df['Percentage'] < 40].shape[0]
        print(f"\nNumber of Fail Students: {fail_students_count}")

    elif choice == '10':
        course_avg_performance = df.groupby('Course')['Percentage'].mean()
        print("\nCourse-wise Average Performance:")
        print(course_avg_performance)

    elif choice == '11':
        city_avg_performance = df.groupby('City')['Percentage'].mean()
        print("\nCity-wise Average Performance:")
        print(city_avg_performance)

    elif choice == '12':
        gender_avg_performance = df.groupby('Gender')['Percentage'].mean()
        print("\nGender-wise Average Performance:")
        print(gender_avg_performance)

    elif choice == '13':
        course_avg = df.groupby('Course')['Percentage'].transform('mean')
        above_avg_students = df[df['Percentage'] > course_avg]
        print("\nStudents Performing Above Their Course Average:")
        print(above_avg_students)

    elif choice == '14':
        df['Enrollment Month'] = pd.to_datetime(df['Enrollment Date']).dt.month
        month_student_count = df['Enrollment Month'].value_counts()
        most_students_month = month_student_count.idxmax()
        print(f"\nMonth with the most new students: {most_students_month} ({month_student_count[most_students_month]} students)")

    elif choice == '15':
        final_analysis = pd.DataFrame({
            'Highest Percentage Student': [df.loc[df['Percentage'].idxmax()]['Name']],
            'Lowest Percentage Student': [df.loc[df['Percentage'].idxmin()]['Name']],
            'Course with Highest Average Percentage': [df.groupby('Course')['Percentage'].mean().idxmax()],
            'City with Highest Average Attendance': [df.groupby('City')['Attendance'].mean().idxmax()],
            'Course with Most Students': [df['Course'].value_counts().idxmax()],
            'Student with Highest Python Marks': [df.loc[df['Python Marks'].idxmax()]['Name']],
            'Average Attendance': [df['Attendance'].mean()],
            'Number of Pass Students': [df[df['Percentage'] >= 40].shape[0]],
            'Number of Fail Students': [df[df['Percentage'] < 40].shape[0]],
        })
        print("\nFinal Analysis:")
        print(final_analysis)

    elif choice == '16':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please enter a number between 1 and 16.")
```
