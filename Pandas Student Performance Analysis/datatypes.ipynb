```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1. Find datatype of every column.")
    print("2. Convert age to integer.")
    print("3. Convert joining_date to datetime.")
    print("4. Verify joining_date datatype.")
    print("5. Create joining_year.")
    print("6. Create joining_month.")
    print("7. Exit.")

    choice = input("Enter your choice (1-6): ")
    print("\n")

    if choice == '1':
        print("\nDatatypes of every column:")
        print(df.dtypes)

    elif choice == '2':
        df['age'] = df['age'].astype(int)
        print("\nAge column converted to integer.")
        print(df.dtypes['age'])

    elif choice == '3':
        df['joining_date'] = pd.to_datetime(df['joining_date'])
        print("\nJoining date column converted to datetime.")
        print(df.dtypes['joining_date'])
        
    elif choice == '4':
        print("\nVerifying joining_date datatype:")
        print(df.dtypes['joining_date'])

    elif choice == '5':
        df['joining_year'] = df['joining_date'].dt.year
        print("\nJoining year column created.")

    elif choice == '6':
        df['joining_month'] = df['joining_date'].dt.month
        print("\nJoining month column created.")

    elif choice == '7':
        print("Exiting the program.")
        break
    
    else:
        print("Invalid choice. Please enter a number between 1 and 6.")
```
