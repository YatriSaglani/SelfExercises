```python
import pandas as pd
```

```python
students = pd.read_csv("student_data.csv")
fees = pd.read_csv("fees.csv")
```

```python
while True:

    print("\n1. Inner Join")
    print("2. Left Join")
    print("3. Right Join")
    print("4. Outer Join")
    print("5. Exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':

        result = pd.merge(
            students,
            fees,
            on="id",
            how="inner"
        )

        print("\n--- Inner Join ---")
        print(result)

    elif choice == '2':

        result = pd.merge(
            students,
            fees,
            on="id",
            how="left"
        )

        print("\n--- Left Join ---")
        print(result)

    elif choice == '3':

        result = pd.merge(
            students,
            fees,
            on="id",
            how="right"
        )

        print("\n--- Right Join ---")
        print(result)

    elif choice == '4':

        result = pd.merge(
            students,
            fees,
            on="id",
            how="outer"
        )

        print("\n--- Outer Join ---")
        print(result)

    elif choice == '5':

        print("Exiting...")
        break

    else:

        print("Invalid choice. Please try again.")
```
