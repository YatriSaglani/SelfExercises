```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1. Sort age ascending.")
    print("2. Sort age descending.")
    print("3. Sort Python marks descending.")
    print("4. Sort percentage descending.")
    print("5. Sort attendance ascending.")
    print("6. Sort course ascending and percentage descending.")
    print("7. Exit.")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':
        print(df.sort_values('age'))

    elif choice == '2':
        print(df.sort_values('age', ascending=False))

    elif choice == '3':
        print(df.sort_values('python', ascending=False))

    elif choice == '4':
        print(df.sort_values('percentage', ascending=False))

    elif choice == '5':
        print(df.sort_values('attendance'))

    elif choice == '6':
        print(df.sort_values(['course', 'percentage'], ascending=[True, False]))

    elif choice == '7':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please try again.")
```
