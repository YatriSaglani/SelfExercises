```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1. Age > 20 students.")
    print("2. Age <= 20 students.")
    print("3. Python course students.")
    print("4. Rajkot students.")
    print("5. Attendance > 90.")
    print("6. Python marks > 80.")
    print("7. Maths marks between 60 and 80.")
    print("8. Python course AND attendance > 85.")
    print("9. Rajkot OR Ahmedabad students.")
    print("10. Percentage >= 75.")
    print("0. Exit.")

    choice = input("Enter your choice (1-10) : ")
    print("\n")

    if choice == '1':
        result = df[df['age'] > 20]
        print(result)

    elif choice == '2':
        result = df[df['age'] <= 20]
        print(result)

    elif choice == '3':
        result = df[df['course'] == 'Python']
        print(result)

    elif choice == '4':
        result = df[df['city'] == 'Rajkot']
        print(result)

    elif choice == '5':
        result = df[df['attendance'] > 90]
        print(result)

    elif choice == '6':
        result = df[df['Python'] > 80]
        print(result)

    elif choice == '7':
        result = df[(df['Maths'] >= 60) & (df['Maths'] <= 80)]
        print(result)

    elif choice == '8':
        result = df[(df['course'] == 'Python') & (df['Attendance'] > 85)]
        print(result)

    elif choice == '9':
        result = df[(df['city'] == 'Rajkot') | (df['City'] == 'Ahmedabad')]
        print(result)

    elif choice == '10':
        result = df[df['percentage'] >= 75]
        print(result)

    elif choice == '0':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please enter a number between 0 and 10.")
```
