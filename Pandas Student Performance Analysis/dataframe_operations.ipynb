```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1. Display name, age and course columns.")
    print("2. Display maths, python, pandas and numpy columns.")
    print("3. Create total = maths + python + pandas + numpy.")
    print("4. Create percentage.")
    print("5. Create result: percentage >= 40 if Pass, otherwise Fail.")
    print("6.exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':
        print(df[['name', 'age', 'course']])

    elif choice == '2':
        print(df[['maths', 'python', 'pandas', 'numpy']])

    elif choice == '3':
        df['total'] = df['maths'] + df['python'] + df['pandas'] + df['numpy']
        print(df[['name', 'total']])

    elif choice == '4':
        df['percentage'] = (df['total'] / 400) * 100
        print(df[['name', 'percentage']])

    elif choice == '5':
        df['result'] = df['percentage'].apply(lambda x: 'Pass' if x >= 40 else 'Fail')
        print(df[['name', 'result']])

    elif choice == '6':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please try again.")
```
