```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\nSelect an option:")
    print("1. Read CSV file")
    print("2. Display first 5 rows")
    print("3. Display last 5 rows")
    print("4. Find total rows and columns")
    print("5. Display column names")
    print("6. Display info()")
    print("7. Display describe() for numerical columns")
    print("8. Exit")

    choice = input("Enter your choice (1-8): ")
    print("\n")

    if choice == '1':
        print(df)

    elif choice == '2':
        print(df.head())

    elif choice == '3':
        print(df.tail())

    elif choice == '4':
        print("Total rows:", df.shape[0])
        print("Total columns:", df.shape[1])

    elif choice == '5':
        print("Column names:")
        print(df.columns.tolist())

    elif choice == '6':
        df.info()

    elif choice == '7':
        print(df.describe())

    elif choice == '8':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please try again.")
```
