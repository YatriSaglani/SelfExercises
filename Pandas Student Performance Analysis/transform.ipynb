```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
df["total"] = df["maths"] + df["python"] + df["pandas"] + df["numpy"]

df["percentage"] = df["total"] / 4

df["result"] = df["percentage"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
```

```python
while True:

    print("\n1. Course-wise Average Python Marks")
    print("2. City-wise Average Percentage")
    print("3. Create Course Average Python Column")
    print("4. Compare Python Marks with Course Average")
    print("5. Exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':
        df["course_avg_python"] = df.groupby("course")["python"].transform("mean")

        print("\nCourse-wise Average Python Marks:")
        print(df[["course", "python", "course_avg_python"]])

    elif choice == '2':
        df["city_avg_percentage"] = df.groupby("city")["percentage"].transform("mean")

        print("\nCity-wise Average Percentage:")
        print(df[["city", "percentage", "city_avg_percentage"]])

    elif choice == '3':
        df["course_avg_python"] = df.groupby("course")["python"].transform("mean")

        print("\nCourse Average Python Column:")
        print(df[["name", "course", "python", "course_avg_python"]])

    elif choice == '4':
        df["course_avg_python"] = df.groupby("course")["python"].transform("mean")

        df["performance"] = df.apply(
            lambda row: "Above Average"
            if row["python"] > row["course_avg_python"]
            else "Below Average",
            axis=1
        )

        print("\nPython Performance:")
        print(df[
            ["name", "course", "python", "course_avg_python", "performance"]
        ])

    elif choice == '5':
        print("Exiting Task 11.")
        break

    else:
        print("Invalid choice. Please try again.")
```
