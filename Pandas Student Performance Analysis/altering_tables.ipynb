```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:

    print("\n1. Rename Columns")
    print("2. Add Grade Column")
    print("3. Delete Gender Column")
    print("4. Change Column Order")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == '1':

        df.rename(columns={
            "name": "student_name",
            "python": "python_marks",
            "pandas": "pandas_marks",
            "numpy": "numpy_marks"
        }, inplace=True)

        print("\nColumns renamed successfully!")
        print(df.columns.tolist())

    elif choice == '2':

        df["percentage"] = pd.to_numeric(df["percentage"])

        df["grade"] = df["percentage"].apply(
            lambda x: "A" if x >= 80
            else "B" if x >= 70
            else "C" if x >= 60
            else "D" if x >= 50
            else "F"
        )

        print("\nGrade column added:")
        print(df[["student_name", "percentage", "grade"]])

    elif choice == '3':

        df.drop("gender", axis=1, inplace=True)

        print("\nGender column deleted successfully!")
        print(df.columns.tolist())

    elif choice == '4':

        new_order = [
            "id",
            "student_name",
            "age",
            "city",
            "course",
            "maths",
            "python_marks",
            "pandas_marks",
            "numpy_marks",
            "attendance",
            "joining_date",
            "percentage",
            "grade"
        ]

        df = df[new_order]

        print("\nColumn order changed:")
        print(df)

    elif choice == '5':

        print("Exiting...")
        break

    else:

        print("Invalid choice. Please try again.")
```
