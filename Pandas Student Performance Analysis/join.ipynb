```python
import pandas as pd
```

```python
student_data = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "name": ["Raj", "Priya", "Dev", "Anjali", "Arjun"],
    "course": ["Python", "Data Science", "AI", "Machine Learning", "Deep Learning"]
})

performance_data = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "percentage": [82, 91, 76, 88, 79],
    "attendance": [92, 95, 87, 97, 90]
})


while True:

    print("\n1. Display Student Data")
    print("2. Display Performance Data")
    print("3. Join Both DataFrames")
    print("4. Exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':

        print("\n--- Student Data ---")
        print(student_data)

    elif choice == '2':

        print("\n--- Performance Data ---")
        print(performance_data)

    elif choice == '3':

        student_data_indexed = student_data.set_index("id")
        performance_data_indexed = performance_data.set_index("id")

        result = student_data_indexed.join(
            performance_data_indexed
        )

        print("\n--- Joined Data ---")
        print(result)

    elif choice == '4':

        print("Exiting...")
        break

    else:

        print("Invalid choice. Please try again.")
```
