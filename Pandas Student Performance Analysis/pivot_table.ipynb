```python
import pandas as pd
```

```python
df["total"] = df["maths"] + df["python"] + df["pandas"] + df["numpy"]

df["percentage"] = df["total"] / 4

df["result"] = df["percentage"].apply(
    lambda x: "Pass" if x >= 40 else "Fail"
)
```

```python
while True:
   
    print("\n1. Course-wise Average Percentage")
    print("2. City-wise Gender Average Percentage")
    print("3. Course-wise Gender Average Python Marks")
    print("4. Exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':
        pivot1 = pd.pivot_table(
            df,
            index="course",
            values="percentage",
            aggfunc="mean"
        )

        print("\nCourse-wise Average Percentage:")
        print(pivot1)

    elif choice == '2':
        pivot2 = pd.pivot_table(
            df,
            index="city",
            columns="gender",
            values="percentage",
            aggfunc="mean"
        )

        print("\nCity-wise Gender Average Percentage:")
        print(pivot2)

    elif choice == '3':
        pivot3 = pd.pivot_table(
            df,
            index="course",
            columns="gender",
            values="python",
            aggfunc="mean"
        )

        print("\nCourse-wise Gender Average Python Marks:")
        print(pivot3)

    elif choice == '4':
        print("Exiting...")
        break

    else:
        print("Invalid choice. Please try again.")
```
