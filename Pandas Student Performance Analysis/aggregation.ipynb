```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1.Calculate average maths marks.")
    print("2.Calculate average python marks.")
    print("3.Calculate maximum python marks.")
    print("4.Calculate minimum python marks.")
    print("5.Calculate total number of students.")
    print("6.Calculate average attendance.")
    print("7.Calculate minimum percentage.")
    print("8.Calculate maximum percentage.")
    print("9.Exit.")

    choice = input("Enter your choice (1-9): ")

    if choice == '1':
        avg_maths = df['maths'].mean()
        print(f"\nAverage Maths Marks: {avg_maths}")

    elif choice == '2':
        avg_python = df['python'].mean()
        print(f"\nAverage Python Marks: {avg_python}")

    elif choice == '3':
        max_python = df['python'].max()
        print(f"\nMaximum Python Marks: {max_python}")

    elif choice == '4':
        min_python = df['python'].min()
        print(f"\nMinimum Python Marks: {min_python}")

    elif choice == '5':
        total_students = df.shape[0]
        print(f"\nTotal Number of Students: {total_students}")

    elif choice == '6':
        avg_attendance = df['attendance'].mean()
        print(f"\nAverage Attendance: {avg_attendance}")

    elif choice == '7':
        min_percentage = df['percentage'].min()
        print(f"\nMinimum Percentage: {min_percentage}")

    elif choice == '8':
        max_percentage = df['percentage'].max()
        print(f"\nMaximum Percentage: {max_percentage}")

    elif choice == '9':
        print("Exiting...")
        break

    else:
        print("Invalid choice. Please enter a number between 1 and 9.")
```
