```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")

df["total"] = df["maths"] + df["python"] + df["pandas"] + df["numpy"]

df["percentage"] = df["total"] / 4

```

```python
while True:
    print("\n1. Course-wise Analysis")
    print("2. City-wise Analysis")
    print("3. Gender-wise Analysis")
    print("4. exit")

    choice = input("Enter your choice: ")
    print("\n")

    if choice == '1':
        course_analysis = df.groupby("course").agg(
            total_students=("id", "count"),
            average_maths=("maths", "mean"),
            average_python=("python", "mean"),
            average_pandas=("pandas", "mean"),
            average_percentage=("percentage", "mean"))

        print(course_analysis)

    elif choice == '2':
        city_analysis = df.groupby("city").agg(
            total_students=("id", "count"),
            average_percentage=("percentage", "mean"),
            average_attendance=("attendance", "mean"))

        print(city_analysis)

    elif choice == '3':
        gender_analysis = df.groupby("gender").agg(
            total_students=("id", "count"),
            average_percentage=("percentage", "mean"))

        print(gender_analysis)          

    elif choice == '4':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please try again.")
```
