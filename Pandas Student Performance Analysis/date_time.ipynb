```python
import pandas as pd
```

```python
df = pd.read_csv("students.csv")
```

```python
while True:
    print("\n1. Extract year.")
    print("2. Extract month.")
    print("3. Sort by joining date.")
    print("4. Filter students who joined after 2026-01-01.")
    print("5. Find students who joined in January.")
    print("6. Exit.")

    choice = input("Enter your choice (1-6): ")
    print("\n")

    if choice == '1':
        df['joining_year'] = pd.to_datetime(df['joining_date']).dt.year
        print("\nJoining year column created.")

    elif choice == '2':
        df['joining_month'] = pd.to_datetime(df['joining_date']).dt.month
        print("\nJoining month column created.")

    elif choice == '3':
        df_sorted = df.sort_values(by='joining_date')
        print("\nDataFrame sorted by joining date:")
        print(df_sorted)

    elif choice == '4':
        filtered_df = df[pd.to_datetime(df['joining_date']) > pd.to_datetime('2026-01-01')]
        print("\nStudents who joined after 2026-01-01:")
        print(filtered_df)

    elif choice == '5':
        january_students = df[pd.to_datetime(df['joining_date']).dt.month == 1]
        print("\nStudents who joined in January:")
        print(january_students)

    elif choice == '6':
        print("Exiting the program.")
        break

    else:
        print("Invalid choice. Please enter a number between 1 and 6.")
```
